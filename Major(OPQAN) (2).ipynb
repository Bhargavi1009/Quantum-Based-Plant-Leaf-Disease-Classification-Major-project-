```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

IMG_SIZE = 224
BATCH_SIZE = 32

train_datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

train_data = train_datagen.flow_from_directory(
    base_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='training'
)

val_data = train_datagen.flow_from_directory(
    base_dir,
    target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE,
    class_mode='categorical',
    subset='validation'
)

NUM_CLASSES = train_data.num_classes
```

```python
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras import layers, models

base_model = MobileNetV2(weights='imagenet', include_top=False, input_shape=(224,224,3))
base_model.trainable = False

x = base_model.output
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dense(256, activation='relu')(x)
x = layers.Dropout(0.3)(x)
output = layers.Dense(NUM_CLASSES, activation='softmax')(x)

model_classical = models.Model(inputs=base_model.input, outputs=output)

model_classical.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

history = model_classical.fit(
    train_data,
    validation_data=val_data,
    epochs=6
)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'])
plt.plot(history.history['val_accuracy'])

plt.title("Model Accuracy")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")

plt.legend(['Train Accuracy', 'Validation Accuracy'])

plt.show()
```

```python
def generate_polynomials(n):
    polys = []
    t = np.arange(1, n+1)
    mu = (n+1)/2

    u0 = np.ones(n)
    u1 = t - mu

    polys.append(u0)
    polys.append(u1)

    for i in range(1, n-1):
        bi = (i**2 * (n**2 - i**2)) / (4*(4*i**2 - 1))
        ui = (t - mu)*polys[i] - bi*polys[i-1]
        polys.append(ui)

    return np.array(polys)


def opt_features(image, size=8):
    img = cv2.resize(image, (size, size))
    img = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

    M = generate_polynomials(size)
    beta = M @ img @ M.T

    return beta.flatten()
```

```python
X_fused = []
y_fused = []

for batch_images, batch_labels in train_data:

    cnn_feats_batch = cnn_extractor.predict(batch_images)

    for img, cnn_feat, label in zip(batch_images, cnn_feats_batch, batch_labels):

        # Safe conversion
        img_np = img

        if img_np.max() <= 1.0:
            img_np = img_np * 255

        img_np = img_np.astype('uint8')

        opt_feat = opt_features(img_np)

        # Normalize
        cnn_feat = cnn_feat.flatten() / (np.linalg.norm(cnn_feat.flatten()) + 1e-8) # Flatten cnn_feat here
        opt_feat = opt_feat / (np.linalg.norm(opt_feat) + 1e-8)

        fused = np.concatenate([cnn_feat, opt_feat])
        X_fused.append(fused)

        label_val = np.argmax(label)
        y_fused.append(label_val)

    break   # VERY IMPORTANT (generator stop)
```

```python
import torch
import torch.nn as nn
import pennylane as qml
import numpy as np
from sklearn.metrics import confusion_matrix

n_qubits = 20 # Changed from 6 to 20 to match PCA output
dev = qml.device("default.qubit", wires=n_qubits)

@qml.qnode(dev, interface='torch')
def quantum_circuit(inputs, weights):

    qml.AngleEmbedding(inputs, wires=range(n_qubits))


    for i in range(n_qubits):
        qml.RY(weights[i], wires=i)
    for i in range(n_qubits - 1):
        qml.CNOT(wires=[i, i + 1])

    return [qml.expval(qml.PauliZ(i)) for i in range(n_qubits)]

class HybridModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.classical = nn.Linear(n_qubits, n_qubits) # Input from PCA is n_qubits (6)

        self.qlayer = qml.qnn.TorchLayer(quantum_circuit, weight_shapes={"weights": (n_qubits,)})
        self.output = nn.Linear(n_qubits, NUM_CLASSES) # FIX: Use actual NUM_CLASSES

    def forward(self, x):
        x = torch.relu(self.classical(x))
        x = self.qlayer(x)
        return torch.softmax(self.output(x), dim=1)

model = HybridModel()
print(model)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

# tensors
X_train_tensor = torch.tensor(X_train, dtype=torch.float32)
y_train_tensor = torch.tensor(y_train, dtype=torch.long)

X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
y_test_tensor = torch.tensor(y_test, dtype=torch.long)

model = HybridModel()

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

epochs = 120

for epoch in range(epochs):
    model.train()

    optimizer.zero_grad()
    outputs = model(X_train_tensor)

    loss = criterion(outputs, y_train_tensor)
    loss.backward()
    optimizer.step()

    # Train accuracy
    _, train_pred = torch.max(outputs, 1)
    train_acc = (train_pred == y_train_tensor).float().mean()

    # Validation accuracy
    model.eval()
    with torch.no_grad():
        val_outputs = model(X_test_tensor)
        _, val_pred = torch.max(val_outputs, 1)
        val_acc = (val_pred == y_test_tensor).float().mean()

    print(f"Epoch {epoch+1}/{epochs} | Loss: {loss.item():.4f} | Train Acc: {train_acc.item():.4f} | Val Acc: {val_acc.item():.4f}")
```
